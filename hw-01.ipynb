```python
import torch
import torchvision.transforms as transforms
import matplotlib.pyplot as plt
from torchvision.io import read_image
```

```python
image_path = "2ed61a1c124c62e813be5dfb3b3f5dbb.jpeg" #Загружаем Сэма
image = read_image(image_path).float() / 255.0
```

# Пайплайн аугментаций

```python
augmentations = [
    ("Resize", transforms.Resize((256, 256))),
    ("RandomHorizontalFlip", transforms.RandomHorizontalFlip(p=1)), #Сделаем так, чтобы флип всегда происходил
    ("ColorJitter", transforms.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.2, hue=0.1)),
    ("RandomRotation", transforms.RandomRotation(degrees=30)),
    ("GaussianBlur", transforms.GaussianBlur(kernel_size=(5, 5), sigma=(0.1, 2.0)))
]
```

```python
def show_image(image, title):
    plt.figure(figsize=(5, 5))
    plt.imshow(image.permute(1, 2, 0))
    plt.title(title)
    plt.axis("off")
    plt.show()
```

# Выводим изображение

```python
transformed_image = image.clone()
show_image(transformed_image, "Оригинал")
```

```python
for name, transform in augmentations:
    transformed_image = transform(transformed_image)
    show_image(transformed_image, name)
```

# Выводы

Были выбраны следующие аугментации:
1) Resize (изменяем размер изображения до 256x256)

2) HorizontalFLip (отражаем изображение по горизонтали)

3) ColorJitter (меняем яркость, контрастность, насыщенность и оттенок)

4) RandomRotation (вращаем на рандомный угол до 30 градусов)

5) GaussianBlur (применяем размытие Гаусса)

Аугментации различных типов.
